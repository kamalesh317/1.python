# Create a function that checks whether the given number is Odd or Even 

```python
Number = int(input("Enter the number :"))
def oddeven():
  if Number % 2 == 0:
    print(Number,"is a Even Number")
  else:
    print(Number,"is a Odd Number")
 
```

```python
oddeven()
```

# calculate the percentage of your 10th mark 

```python
print("Enter the 10th marks to calculate your percentage")
Subject1 = int(input("Subject1  :"))
Subject2 = int(input("Subject2  :"))
Subject3 = int(input("Subject3  :"))
Subject4 = int(input("Subject4  :"))
Subject5 = int(input("Subject5  :"))
def percentage ():
    Total = Subject1 + Subject2 + Subject3 + Subject4 + Subject5
    percentage = Total / 5
    print("Your total mark is :",Total)
    print("Your percentage is :",percentage)

```

```python
percentage()
```

# print area and perimeter of triangle using class and functions

```python
Height = int(input("Height: " ))
Breadth = int(input("Breadth: "))
def area():
    Areaformula = (Height*Breadth)/2
    print("Area of Triangle",Areaformula)
def perimeter():
    Height1 = int(input("Height1 : "))
    Height2 = int(input("Height2 : "))
    Breadth = int(input("Breadth : "))
    Perimeterformula = Height1 + Height2 + Breadth
    print("Perimeter of the triangle",Perimeterformula)
area()
perimeter()
```
